```python
student={
    "Name":"Ram Das",
    "Father Name": "Thomas",
    "Age":23,
    "Course":"BCA",
    "Age":35,
    "Roll No":78
}
print(student)
print(len(student))
mykeys=student.keys()
print(mykeys)
```

```python
myvalues = student.values()
print(myvalues)
```

```python
mykeys=student.keys()
print(mykeys)
```

```python
myitems=student.items()
print(myitems)
```

```python
#changing dictionaries items
car={
    "Brand":"Ford",
    "Model":"Mustang",
    "Year":1964
}
car["Year"]=2023
print(car)
```

```python
#adding dictionaries items
car={
    "Brand":"Ford",
    "Model":"Mustang",
    "Year":1964
}
car["Color"]="Red"
print(car)
```

```python
#removing dictionaries items
car={
    "Brand":"Ford",
    "Model":"Mustang",
    "Year":1964
}
car.pop("Model")
print(car)
```
