```python
#multiline plot
import matplotlib.pyplot as plt
import numpy as np
y1 = np.array([4,7,3,9])
y2 = np.array([7,2,5,9])
plt.plot(y1)
plt.plot(y2)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
xpoints = np.array([20,25,35,45])
ypoints = np.array([60,75,56,85])
plt.plot(xpoints,ypoints)

plt.title("My Burnage Calories")
plt.xlabel("Duration")
plt.ylabel("calories")
plt.show()
```

```python
y1 = np.array([2,7,3,9])
y2 = np.array([4,2,7,5])
plt.subplot(1,2,1)
plt.plot(y1)

#plot 2
plt.subplot(1,2,2)
plt.plot(y2, c="red")
plt.show()
```

```python
#scatter plot
x=np.array([2,5,7,8,9,4])
y=np.array([10,15,20,2,11,4])
plt.scatter(x,y)
plt.show()
```

```python
x=np.array([2,4,6,3,9])
y=np.array([11,3,7,8,5])

x1=np.array([1,3,7,8,6])
y1=np.array([3,5,7,8,11])

plt.scatter(x,y)
plt.scatter(x1,y1)
plt.show()
```

```python
x=np.array([2,4,6,3,9])
y=np.array([11,3,7,8,5])

x1=np.array([1,3,7,8,6])
y1=np.array([3,5,7,8,11])

plt.scatter(x,y,c="red")
plt.scatter(x1,y1,c="#00ff00")
plt.show()
```

```python
x=np.array([2,5,7,8,9,4])
y=np.array([10,15,20,2,11,4])
c=np.array(["blue","green","red","black","brown","orange"])
plt.scatter(x,y,color=c)
plt.show()
```

```python
#barh
x=np.array(["A","B","C","D"])
y=np.array([20,2,11,4])
c=np.array(["blue","green","red","black"])
plt.barh(x,y,color=c)
plt.show()
```

```python
#bar
x=np.array(["A","B","C","D"])
y=np.array([20,2,11,4])
c=np.array(["blue","green","red","black"])
plt.bar(x,y,color=c)
plt.show()
```

```python
#pie chart with labels
y = np.array([35,25,25,15])
label=["Apple","lemon","Orange","Kiwi"]
plt.pie(y,labels=label)
plt.show()
```

```python
#pie chart without labels
y = np.array([35,25,25,15])

plt.pie(y)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
y = np.array([35,25,25,15])
mylabels = ["Apple","Banana","Cherries","Dates"]
myexplode = [0.2,0,0,0]
plt.pie(y,labels = mylabels,explode=myexplode)
plt.show()
```
