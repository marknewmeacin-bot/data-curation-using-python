```python
myTuple=("apple","banana","cherry")
print("Original TUple:", myTuple)
mylist=list(myTuple)
mylist[1]="kiwi"
print("Modify List:",mylist)
newTuple=tuple(mylist)
print("newTuple:",newTuple)
```

```python
#adding an item to a tuple
myTuple=("C#","java","React")
mylist=list(myTuple)
mylist.append("Python")
newTuple = tuple(mylist)
print(newTuple)
```

```python
#adding a tuple to another tuple
tuple1=("Apple","lemon")
tuple2=("orange","kiwi")
tuple1 +=tuple2
print(tuple1)
```

```python
#unpaking tuple
fruits=("Grapes","Banana","Apple")
(green,yellow,red)=fruits
print(green)
print(yellow)
print(red)
```

```python
#unpaking tuple
fruits=("Grapes","Banana","Apple","Cherry","kiwi")
(green,yellow,*red)=fruits
print(green)
print(yellow)
print(red)
```
