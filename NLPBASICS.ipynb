```python
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
#Sample text
text = "NLTK is a powerful tool for Natural Language Preocessing. I love to learn NLP."
```

```python
#Tokenization
print("Word Tokenization:", word_tokenize(text))
print("Sentence Tokenization:", sent_tokenize(text))
```

```python
#Stop Words Removal
stop_words= set(stopwords.words('english'))
words= word_tokenize(text)
filtered_words = [word for word in words if word.lower()not in stop_words]
print("Filtered Words:", filtered_words)
```

```python
#Stemming and Lemmatization
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
print("Stemming:",[stemmer.stem(word) for word in filtered_words])
print("Lemmatization:",[lemmatizer.lemmatize(word,pos='v')for word in filtered_words])
```

```python
from nltk import pos_tag, word_tokenize
from nltk.chunk import ne_chunk

#sample text
text="Barack Obama was the 44th President of the United States."
```

```python
#POS Tagging
tokens = word_tokenize(text)
pos_tags = pos_tag(tokens)
print("POS Tags:", pos_tags)
```

```python
#chunking
chunks = ne_chunk(pos_tags)
print("Named Entity Recognition:", chunks)
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.decomposition import LatentDirichletAllocation
```

```python
#Sample corpus
corpus = ["AL is transforming industries.","NLP is a branch of AL.","Text analysis is a part of NLP"]
```

```python
#Bag of Words
vectorizer= CountVectorizer()
x = vectorizer.fit_transform(corpus)
print("BoW Representation:", x.toarray())
print("Feature Names:",vectorizer.get_feature_names_out())
```

```python
#Topic Modeling with LDA
lda = LatentDirichletAllocation(n_components=2,random_state=0)
lda.fit(x)
print("Topics:",lda.components_)
```
